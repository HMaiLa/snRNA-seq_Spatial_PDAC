```python
# PLOTTING THOSE CELLS WITH MUTATION 
import pandas as pd
import numpy as np

# Extract expression values
expr = adata_106_480[:, "NM_004985_ALT_34:A"].X
if hasattr(expr, "toarray"):  # handle sparse matrix
    expr = expr.toarray().flatten()
else:
    expr = np.ravel(expr)

# Combine with metadata
df = adata_106_480.obs.copy()
df["NM_004985_ALT_34:A_expr"] = expr
df["is_expressed"] = df["NM_004985_ALT_34:A_expr"] > 0
```
